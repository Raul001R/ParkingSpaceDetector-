```python
    img = cv2.imread("lot.jpg")
    counter = 0
    for box in results[0].boxes:

        if int(box.cls[0]) == 2:
            x1, y1, x2, y2 = map(int, box.xyxy[0])
            cv2.rectangle(img, (x1, y1), (x2, y2), (0, 255, 0), 2)
            cv2.putText(img,f"Car {counter} confidence of {round(box.conf[0].item(), 2)}", (x1, y1 - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.9, (36,255,12), 2)
            counter += 1
            
    print(f"{counter} cars detected")
    cv2.imwrite("lot_with_boxes.jpg", img)
```
